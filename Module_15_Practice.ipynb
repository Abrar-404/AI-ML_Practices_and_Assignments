# Q1.
## Suppose you have the sentence: "I love machine learning today"
### Using a Skip-gram model with window size = 1:
- What are the input-target pairs generated?
- Which word is the input when predicting "I"?
- Which word is the input when predicting "machine"?


```python
words = "I love machine learning today".split()
pairs = []

for i in range(len(words)):
  if i - 1 >= 0:
    pairs.append((words[i], words[i - 1]))

  if i + 1 < len(words):
    pairs.append((words[i], words[i + 1]))


# What are the input-target pairs generated?
print("Input-Target Pairs:")
print(pairs)

# Which word is the input when predicting "I"?
print('\nInputs for I:', [a for a, b in pairs if b == 'I'])

# Which word is the input when predicting "machine"?
print("\nInputs for 'machine':", [a for a, b in pairs if b == "machine"])
```

# Q2.
## Why do we use Word2Vec instead of One-Hot Encoding?
### Explain in terms of:
- Dimensionality
- Similarity between words
- Memory usage


We use Word2Vec instead of OneHotEncoding for following reasons:

- Dimentionality:
  - **One-Hot Encoding**: Creates huge vectors equal to the total vocabulary size (e.g., 50,000 dimensions for 50,000 words).

  - **Word2Vec**: Uses compact, fixed-size dense vectors (usually 100 to 300 dimensions) no matter how large the vocabulary is.

- Similarity between words:

  - **One-Hot Encoding**: Treats every word as completely independent. Words like "cat" and "dog" have zero relationship in this system.

  - **Word2Vec**: Captures meaning and context, placing words with similar meanings (like "cat" and "dog") close together in vector space.

- Memory usage:

  - **One-Hot Encoding**: Wastes memory because vectors are mostly zeros (sparse).

  - **Word2Vec**: Uses dense, efficient numbers that save significant memory.

# Q3.
## What is the difference between:
- CBOW
- Skip-gram
- Which one predicts context from a target word?


- CBOW (Continuous Bag of Words):
  - Predicts a target word using its surrounding context words.
  - Example: Given the context "The [___] sat on the mat", it predicts "cat"
  - Speed: Fast to train and works well with large datasets.


- Skip-gram:
  - Predicts the surrounding context words given a single target word.
  - Example: Given the target word "cat", it predicts context words like "The", "sat", and "mat".
  - Performance: Better at handling rare words and smaller datasets.

- Which one predicts context from a target word?
  - Skip-gram is the one that predicts context from a target word.
