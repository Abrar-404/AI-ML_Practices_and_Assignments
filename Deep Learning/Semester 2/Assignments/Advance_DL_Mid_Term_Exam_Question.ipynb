## Explain the concept of Convolutional Neural Networks (CNNs) in detail. Describe why CNNs are preferred over fully connected Artificial Neural Networks for image data, how local connectivity and weight sharing work, and how these ideas help in extracting spatial features from face images used for age-group detection.


# Write Answer 01:

### Convolutional Neural Networks (CNN) in detail:

- Convolutional Neural Networks (CNN) is a type of deep learning algorithm which is mainly used to analyze and process visual data like images and videos. Instead of treating the visual data (images, videos) as a flat list of numbers, cnn analyze small patches of the data at a time and learn to recognize patterns such as edges, textures, shapes etc and then combine them into high level features likes eyes, ears, nose etc.

  **The convolution operation in details**:
    It works by sliding a small matrix of weights called kernel over the image, one position at a time. At each position it computes the dot products between the kernel's weights and the pixel value is currently overlapping and produces a single number. This value is placed into an output called feature map. It repeats the whole process across the whole image and produces the complete feature map for that kernel.

### CNNs are preferred over fully connected Artificial Neural Networks for some reasons.

1. Too many parameters:
   - In a fully connected ANN every pixel is connected to every neuron in the next layer. For example: for a 100 * 100 gray scale image, which is connected to 1000 neurons. It will require 10 million weights in a single layer which will make training very slow, require heavy memory and might tend to overfit.

2. Spatial Structure Loss:
   - A fully connected ANN flatten an image into a 1D vector which generally ignores the pixel positions which are related to each other. As a result it can't recognize the neighboring pixels and also can't recognize the same patterns if these appears in different locations in the image.

3. No translation Invariance:
   - If a pattern is shifted a little bit within the image, the ANN will treat it as a completely different input as the exact pixel positions have changed.

### CNN overcome these problems using local connectivity and weight sharing.

- Local Connectivity:
  - It works by connecting each neuron in CNN only to a local region of the input which is called as receptive field (3 * 3, 5 * 5) instead of connecting the neuron to the entire image. Local connectivity is generally based on the idea that the meaningful patterns such as edges are local. It drastically reduces the number of connections and parameters.

    For example: to detect an eye's edge, it only requires to examine small patch around the eye region, not the entire face.

- Weight Sharing:
  - A CNN uses a small matrix of weights called Kernel which slides across entire image applying the same set of weights at every location. In this, the number of parameters stays small regardless of the image size because the same kernel is used everywhere. This provides translation invariance to the network by which it can detect any feature regardless of the position it appears to.

    For example: a vertical edge kernel will produce a strong response whenever a vertical edge occurs in the image such as near nose, eye or ear because the same kernel is being applied repeatedly.

### These ideas help in extracting spatial features from face images used for age-group detection using a hierarchy of layers.

1. **Convolution Layers**: These helps to detect local patterns such as edges, textures etc.

2. **Activation Fucntion (ReLU)**: It helps to identify non linearity to learn complex patterns.

3. **Pooling Layers**: It downsamples feature maps, retain the strongest activations and reduce computation.

4. **Fully Connected Layers**: It combines extracted high level features to make final classification

The feature abstraction process increases with depths. The early layers helps to detect edges, textures, the middle layers identify textures, curves and small parts and the deep layers identify complex, object specific structures.

The age related cues are inherently local and spatial which makes cnn well suited for the task.

1. The local connectivity allow the early kernel to focus on small, relevant regions where age related features are concentrated.

2. The weight sharing ensures the same feature detectors are applied across the entire image rather than the exact position.

3. The deeper layer combine these local cues into higher level representations which the fully connected layers use to classify the face into age group.

### The architecture of a basic CNN in detail:
A basic CNN processes an image though a sequence of layers, each of which transforms the data step by step from raw pixels to a final class prediction.

1. **Convolutional Layer**:
   - This layer applies kernel that slide all over the image, computes the dot products to detect local patterns like edges, corners, textures etc. Each filter produces a feature map which highlights where the patterns appear.

     For example: One filter might detect horizontal edges, another filter might detect curves.

2. **Activation Fucntions (ReLU)**:
   - ReLU is applied to introduce non linearity by converting all negative values into zero and keeping positive values unchanged. Without this, the network would only learn simple linear relations no matter how many layers it has.

     For example: It helps the network learn complex patterns like wrinkled skin texture.

3. **Pooling Layer**:
   - This layer reduces the size of feature maps by keeping only the strongest values in small regions. This reduces computation and makes the network more tolerant to small shifts in the image.

     For example: If an eye moves a few pixels, pooling still captures that the eye's edge feature was detected somewhere in that region.

4. **Flatten Layer**:
   - This layer converts the final feature maps (2d, 3d) into a single 1d vector so it can be fed into a fully connected layer.

     For example: A 7 * 7 * 64 feature map is flattened into a vector of 3136 values.

5. **Fully Connected Layer**:
   - It takes the flattened feature vector and combine all extracted features to make the final decision
     
     For example: Classifying the face into an age group.

### How increasing the number of convolutional layers and filters affects feature extraction, model performance, and overfitting:

1. **Better feature extraction**:
   - More layers allow the network to learn increasingly abstract features such as the early layers detect edges, the deeper layers detect complex structures like full facial patterns. More filters per layer allow it to detect a wider variety of patterns at each stage.

2. **Improved Performance**:
   - More layers and filters generally improve accuracy because the model can capture more detailed  patterns needed to distinguish age groups.

3. **Overfitting Risks**:
   - After a certain point adding too many filters increases the number of parameters. If the training data is limited the model may memorize training examples rather than learning general patterns which leads to poor performance on unseen face images.

4. **High computational cost**:
   - Adding more layers and filters might also mean slower training, heavy memory usage without providing a guaranteed accuracy gains beyond a certain depth.

  For example: If a filter is designed to detect vertical edges its feature map will show high values wherever a vertical edge exists in the image.

## Explain padding and stride in CNNs in detail. Then explain the convolutional output size formula and what each symbol in the formula means. You do not need to derive the formula. Show how the output height and width change when you use valid padding, same padding, or a larger stride. Also explain why padding is useful when working with cropped face images.


# Write Answer 05:

### Padding and stride in CNNs in detail:

- **Padding**:
  - Padding means adding extra pixels around the border of an image before convolution. Without it the edge pixels are covered by the kernel fewer times than center pixels so the border information is used less and the output shrinks after every layer. There are 2 types of padding.

  1. **Valid Padding**: No pixels are added in this padding and for that the output becomes smaller than input.

  2. **Same Padding**: Zeroes are added so that the output stays the same size as the input.

- **Stride**:
  - It refers to how many pixels the kernel moves in each steps.

  1. If stride = 1: kernel moves one pixel at a time which gives larger and detailed output

  2. If stride = 2: kernel moves two pixel at a time which gives smaller and less detailed output but do the computations faster.


### Output Size Formula:
  - Output = ((input - kernel + 2 * p) / s) + 1

     where,

    - Input = height, width of the image
     - Kernel = height, width of the filter
     - Padding (P) = pixels added to each side
     - Stride (S) = step size of the kernel
     - Output = height, width of the resulting feature map

### Show how the output height and width change when you use valid padding, same padding, or a larger stride:

For Example: Input = 32 * 32, Kernel = 3 * 3

- For stride = 1,	Padding = 0	, (32 - 3 + 0) / 1 + 1 = 30 * 30

- For stride = 1,	Padding = 1	, (32 - 3 + 2) / 1 + 1 = 32 * 32

- For stride = 2,	Padding = 0	, (32 - 3 + 0) / 2 + 1 = 15 * 15

So valid padding shrinks the output, same padding keeps it unchanged and a larger stride shrinks it faster.

### Why Padding Helps with Cropped Face Images:

- In cropped faces the important features like the jaw, ear, hairline often sit near the image border. Without padding the kernel can't fully cover these edge areas so such features get under detected and after several layers the feature map may shrink so much that border details are lost. Same padding gives border regions equal treatment as the center so these edge located facial features are preserved and passed on to deeper layers.

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity when training an age-detection CNN on thousands of face images.


# Write Answer 06:

### Pooling in CNN:
- Pooling is a layer in a CNN that reduces the size of feature maps while keeping the most important information. It works by sliding a small window over the feature map and summarizing the values inside that window into a single number. This reduces the amount of data passed to the next layer without losing the key patterns.

### Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction:
1. **Operation**:
   - Max Pooling: It takes the maximum value from each window.
   - Average Pooling: It takes the mean value of all values from each window.

     For example: For a 2 * 2 window with values 8, 3, 5, 2: max pooling output is 8 and average pooling output is (8 + 3 + 5 + 2) / 4 = 4.5

2. **Advantages**:
   - Max pooling: It keeps the strongest activation so its better at preserving sharp, distinct features like edges or wrinkles which are very useful when detecting whether a specific pattern is present.

   - Average Pooling: Smooths out the feature map by considering all values which reduces noise but can weaken strong, distinct signals.

3. **Impact on Feature Extraction**:
   - Max Pooling: It tends to highlight the most prominent feature in a region which makes it good for detecting presence of specific patterns.
   - Average Pooling: It gives a more general, smoothed summary of a region which can be useful for overall texture information but may blur out sharp details.


### How Pooling Reduces Overfitting and Computational Complexity:
1. **Reduces feature map size**: Pooling shrinks the width and height of feature maps which means that the fewer values are passed to later layers.

2. **Fewer parameters in later layers**: As the flattened vectors which are going into the fully connected layer becomes smaller the number of weights needed there is also reduced which lowers the risk of overfitting on limited data.

3. **Faster training**: With smaller feature maps there are fewer computations needed at each subsequent layer which makes the training faster and less memory intensive which is important when training on thousands of face images.

4. **Some translation tolerance**: Pooling summarizes a small region rather than an exact pixel location so that the small shifts in facial position don't drastically change the output which helps the model generalize better instead of memorizing exact pixel positions.

````python
import os
import torch
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import torch.nn as nn
import torch.optim as optim
from torchvision import transforms
from torch.utils.data import Dataset,DataLoader,Subset
from PIL import Image
import kagglehub

# Part A: Load the data
# Download / unzip the dataset.
path = kagglehub.dataset_download('arashnic/faces-age-detection-dataset')
print('Path:', path)
print()

torch.manual_seed(42)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using Device: {device}')
print()

# Read the CSV file and connect each image ID to its age-group label.
train_csv = os.path.join(path, 'faces', 'train.csv')
img_dir = os.path.join(path, 'faces','Train')

df = pd.read_csv(train_csv)
print(df.head())
print(df['Class'].value_counts())
print()

# Show a few sample images with their labels.
fig, axes = plt.subplots(1, 5, figsize = (10, 8))

for i, ax in enumerate(axes):
  img_id = df.iloc[i]['ID']
  labels = df.iloc[i]['Class']
  img_path = os.path.join(img_dir, img_id)
  img = Image.open(img_path).convert('RGB')
  ax.imshow(img)
  ax.set_title(labels)
  ax.axis('off')

plt.show()

# Part B: Preprocess the images
# Resize all images to the same size (for example 64×64 or 128×128). Convert them to tensors, scale pixel values (normalization), and encode the labels Young, Middle, Old as 0, 1, 2.
class_idx = {'YOUNG': 0, 'MIDDLE': 1, 'OLD': 2}
df['label'] = df['Class'].str.upper().map(class_idx)

#
transform = transforms.Compose(
    [
        transforms.Resize((64, 64)),
        transforms.ToTensor(),
        transforms.Normalize(mean = [0.5] * 3, std = [0.5] * 3)
    ]
)

class FaceDataset(Dataset):
  def __init__(self, dataframe, image_dir, transform = None):
    super().__init__()
    self.df = dataframe.reset_index(drop = True)
    self.image_dir = image_dir
    self.transform = transform

  def __len__(self):
    return len(self.df)

  def __getitem__(self, idx):
    img_id = self.df.iloc[idx]['ID']
    label = self.df.iloc[idx]['label']
    img_path = os.path.join(self.image_dir, img_id)
    image = Image.open(img_path).convert('RGB')

    if (self.transform):
      image = self.transform(image)
    return image, label


dat_full = FaceDataset(df, img_dir, transform)
num_classes = 3
print()
print('Total Samples:', len(dat_full))
print()

# Part C: Split and batch
# Split the data into training and validation sets.
train_dat = int(0.8 * len(dat_full))
val_dat = len(dat_full) - train_dat

train_dataset, val_dataset = torch.utils.data.random_split(dat_full, [train_dat, val_dat])

print('Training Samples:', len(train_dataset))
print('Validation Samples:', len(val_dataset))

# Create DataLoaders so that the model receives images in small batches.
pin = True if device.type == 'cuda' else False
train_loader = DataLoader(train_dataset, batch_size = 32, shuffle = True, pin_memory = pin)
val_loader = DataLoader(val_dataset, batch_size = 32, shuffle = False, pin_memory = pin)

# Part D: Build the CNN
# Build a CNN with four convolutional blocks.
# Each block should have two convolution layers, then activation and normalization, and a pooling layer at the end.
# Increase the number of filters across blocks. After the convolutional part, flatten the feature maps and add fully connected layers with dropout.
# The final layer must have 3 outputs.

class myCNN(nn.Module):
  def __init__(self, num_classes):
    super().__init__()

    self.features = nn.Sequential(
        # 1st block
        nn.Conv2d(3, 32, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.Conv2d(32, 32, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.BatchNorm2d(32),
        nn.MaxPool2d(2),

        # 2nd block
        nn.Conv2d(32, 64, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.Conv2d(64, 64, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.BatchNorm2d(64),
        nn.MaxPool2d(2),

        # 3rd block
        nn.Conv2d(64, 128, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.Conv2d(128, 128, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.BatchNorm2d(128),
        nn.MaxPool2d(2),

        # 4th block
        nn.Conv2d(128, 256, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.Conv2d(256, 256, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.BatchNorm2d(256),
        nn.MaxPool2d(2),
    )

    self.classifier = nn.Sequential(
      nn.Flatten(),
      nn.Linear(256 * 4 * 4, 256),
      nn.ReLU(),
      nn.Dropout(0.4),
      nn.Linear(256, 64),
      nn.ReLU(),
      nn.Dropout(0.4),
      nn.Linear(64, num_classes)
    )

  def forward(self, x):
    x = self.features(x)
    x = self.classifier(x)
    return x

model = myCNN(num_classes = num_classes).to(device)

# Part E: Train and evaluate
# Choose a loss function and an optimizer.
learning_rate = 0.001
epochs = 10
loss_func = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr = learning_rate)

# Train the model
train_losses = []
train_accu = []

for epoch in range(epochs):
  model.train()

  total_loss = 0
  correct = 0
  total = 0

  for batch_features, batch_labels in train_loader:
    batch_features = batch_features.to(device)
    batch_labels = batch_labels.to(device)

    output = model(batch_features)
    loss = loss_func(output, batch_labels)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    total_loss += loss.item()
    _, predicted = torch.max(output, 1)
    total += batch_labels.size(0)
    correct += (predicted == batch_labels).sum().item()

  train_loss = total_loss / len(train_loader)
  train_acc = correct / total
  train_losses.append(train_loss)
  train_accu.append(train_acc)

  print(f'Epoch {epoch+1}/{epochs}, Loss: {train_loss:.4f}, Accuracy: {train_acc:.4f}')

# Evaluation
def evaluate(loader):
  model.eval()

  total_loss = 0.0
  total = 0
  correct = 0

  with torch.no_grad():
    for batch_features, batch_labels in loader:
      batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)

      output = model(batch_features)
      loss = loss_func(output, batch_labels)
      total_loss += loss.item()
      _, predicted = torch.max(output, 1)
      total += batch_labels.size(0)
      correct += (predicted == batch_labels).sum().item()

  return total_loss / len(loader), correct / total

val_loss, val_acc = evaluate(val_loader)
print(f'Val Loss: {val_loss:.4f}, Val Accuracy: {val_acc:.4f}')

#  plot or print training/validation loss or accuracy
fig, axes = plt.subplots(1, 2, figsize = (12, 8))

axes[0].plot(train_losses, label = 'Train loss')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('Training loss curve')
axes[0].legend()

axes[1].plot(train_accu, label = 'Train accuracy')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].set_title('Training accuracy curve')
axes[1].legend()

plt.show()
````
