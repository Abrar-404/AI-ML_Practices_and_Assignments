## What happens if we stack multiple linear layers without activation?

- Without activation function the output will show only the linear data results which means it fails to learn complex patterns despite of how many layers or data is stacked behind. It will only learn the straight line patterns, fail to learn complex patterns like images, videos etc.

## What kind of decision boundary does an MLP create?

- MLP creates non-linear, highly complex decision boundaries. This happens because in MLP there are several hidden layers combined with activation functions working to generate an output from complex, non-linear data or input. The hidden layers with the help of activation functions helps to draw complex decision boundaries to separate and handle complex data and create accurate results.

## What are the main components of a neural network?

- Inputs, weights, bias, hidden layers, activation functions, output layer

## What is the difference between input layer, hidden layer, and output layer?

- The input layer provides data as input to the perceptron or neurons
- The hidden layers with activation functions handle and calculate complex data to prepare output or predictions
- The output layers prints or shows the ultimate result.

## What is forward propagation?

- Forward propagation is the process where input data flows through the network, layer by layer, to produce an output. At each node, the inputs are multiplied by their weights, summed together with a bias, and passed through an activation function. This result is then passed forward as input to the next layer — continuing through all hidden layers until it reaches the output layer, which produces the final prediction.
