# Bias
###1. What is a bias in a neural network?

###2. Why do we add bias to a perceptron?




### 1. What is a bias in a neural network?
 - Ans: Bias in a neural network is a value that allows a model to shift or adjust the output of a neuron. Imagine a scenario where a neuron receives input data. It processes this data by applying weights, representing how important each input is to the overall prediction and then adds the bias. This bias helps the neuron determine whether to “fire” (activate) or not.


###2.  Why do we add bias to a perceptron?
 - Shifting the Decision Boundary:
   - When a neural network tries to separate or classify data, it creates a decision boundary, a line that separates one class from another. Without bias, this boundary could be overly rigid and fail to capture subtler patterns in the data. Bias allows the network to shift and rotate this decision boundary, making it more flexible and capable of handling complex, non-linear problems.
  
 - Filling in the Gaps:
   - Sometimes, a dataset doesn’t provide enough information for a straightforward prediction. Bias helps the model fill in these gaps. It adjusts the output even when certain inputs are missing or ambiguous, allowing for a more accurate and comprehensive model.

- Improving Generalization:
  - Generalization refers to the model’s ability to perform well on unseen data. Without bias, a neural network might perform well on training data but struggle when applied to new scenarios. Bias allows the network to generalize better by giving it the ability to shift predictions in ways that align with unseen patterns.

- Balancing Non-Zero Outputs:
  - If bias didn’t exist, the neural network could become overly dependent on the specific inputs it receives, and outputs could be restricted to narrow ranges. Bias helps ensure that outputs can be shifted appropriately, preventing neurons from being limited to a single starting point. This becomes particularly valuable when dealing with complex, multi-dimensional data where certain inputs are much more critical than others.

# Weights

### 1. What are weights in a perceptron?
### 2. How do weights affect the output of a perceptron?

###1. What are weights in a perceptron?
 - Ans: Weights in a perceptron are numerical parameters that multiply incoming input features to determine how much influence each feature has on the final output decision.

###2. How do weights affect the output of a perceptron?
 - Ans: Weights determine the importance and impact of each input feature, directly scaling the weighted sum that drives the perceptron's final activation output

   - Amplification and Scaling: Each input \(x{i}\) is multiplied by its corresponding weight w{i} (W{i} * X{i}). Higher absolute weight values mean those specific inputs have a greater influence on the internal sum.


   - Positive vs. Negative Impact:
     - A positive weight drives the weighted sum higher, pushing the model toward activating (outputting 1).
     - A negative weight pulls the sum lower, working against activation (outputting 0).
     - A weight of zero neutralizes an input, meaning it contributes nothing to the final choice.
    
   - Geometric Influence: Mathematically, weights define the slope and orientation of the decision boundary (hyperplane) in the input feature space. They tilt or rotate the boundary line to best separate distinct classes of data.

   - Threshold Trigger: The total sum of all weighted inputs plus the bias is passed through an activation function (like a step function). If the weights make the net sum cross the threshold, the perceptron fires.

# Step function

### What is the step function and how does it work?

Ans: A step function (also known as the Heaviside step function) is the simplest activation function in a neural network, acting like a digital switch that outputs either 0 or 1 depending on whether the input crosses a set threshold



How it works: The step function makes a hard "yes or no" choice about whether a neuron should fire.
- Calculate the Input: The neuron takes the sum of all incoming inputs multiplied by their weights, plus a bias value (z = sum(w * x) + b\)
- Compare to Threshold: It checks if this total sum (z) meets or beats a specific threshold limit (usually set to 0).
- Assign Binary Output:
  - If input < 0, output = 0 (the neuron stays off).
  - If input ≥ 0, output = 1 (the neuron turns on / fires)

### Why do we need an activation function in a perceptron?
- Ans: We need an activation function in a perceptron to introduce non-linearity, allowing the network to learn and represent complex patterns that a straight-line linear combination cannot solve

### Name three common activation functions.
- Tanh (Hyperbolic Tangent): This one is kind of like the sigmoid function but just shifted and scaled. It squashes input into a range between -1 and 1
- ReLU (Rectified Linear Unit): outputs the input if positive, otherwise 0: f(x) = max(0, x)
- Softmax: The softmax function transforms the given values into numbers that range from 0 to 1
