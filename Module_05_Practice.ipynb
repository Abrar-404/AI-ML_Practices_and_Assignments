## Explain why sigmoid is called a “squashing function”.

- Sigmoid takes any real-valued input, from − ∞ −∞ to + ∞ +∞, and compresses it into a fixed output range between 0 and 1. Large positive inputs get pushed close to 1, and large negative inputs get pushed close to 0, no matter how extreme the input is. This compressing of an unbounded range into a bounded one gives the function its characteristic S-shape. That squeezing behavior is exactly why it's called a "squashing function."

## Why is ReLU more popular than Sigmoid in deep neural networks?

- ReLU avoids the vanishing gradient problem because its gradient stays constant (1) for positive inputs, unlike sigmoid, whose gradient shrinks to near-zero for large inputs. It's also computationally cheaper since it's just a simple threshold ( max ⁡ ( 0 , 𝑥 ) max(0,x)) with no expensive exponential calculation. ReLU produces sparse activations by outputting exactly 0 for negative inputs, which improves efficiency. Together, these make ReLU converge faster in deep networks, while sigmoid is now mostly used only in output layers for binary classification.

## Why is Binary Cross Entropy preferred over MSE for classification?

- When the prediction is wrong, MSE gives small gradient but BCE gives bigger gradient that helps to correct the error faster

- BCE works with classification problem but MSE works well with regression problem with continuous values.

- Using MSE with Sigmoid can create vanishing gradient problem which BCE ignores.

- BCE is best for calculation probabilities and MSE is for regression problems.

## If true label y = 1 and predicted probability ŷ = 0.9, calculate the Binary Cross Entropy loss.

- BCE = - (y * log(ŷ) + (1 - y) * log(1 - ŷ))
      
      = - (1 * log(0.9) + 0 * log(1 - 0.9))
      = - (-0.1054 + 0)
      = 0.1054

## If y = 0 and ŷ = 0.8, calculate the loss.

- BCE = - (y * log(ŷ) + (1 - y) * log(1 - ŷ))
      
      = - (0 * log(0.8) + 1 * log(1 - 0.8))
      = - (0 - 1.60)
      = 1.60
