# 2. **Lowercasing**

### Consider the text:
### PHITRON is Amazing! I love studying NLP at PHITRON.


### Question:
  - ***Why do we perform lowercasing?***

    ***Ans:*** We perform lowercasing to ensure words with different capitalizations (e.g., "Phitron" and "phitron") are treated as the exact same word.

  - ***What will be the output after lowercasing?***

    ***Ans:*** `phitron is amazing! i love studying nlp at phitron.`

  - ***How does lowercasing help reduce vocabulary size?***

    ***Ans:*** It reduces vocabulary size by merging duplicate words of different cases into single unique tokens.


# 3. **HTML Tag Removal**

### Consider the text:
### <p>Welcome to <b>PHITRON</b>!</p>


### Question:
  - ***Which regex pattern can remove HTML tags?***

    ***Ans:*** `<.*?>`

  - ***Explain why the pattern <.*?> works.***

    ***Ans:*** It works by matching an opening angle bracket, any characters inside it, and a closing angle bracket.

  - ***What is the purpose of ? in this regex?***

    ***Ans:*** The `?` makes the match non-greedy, meaning it stops at the first closing bracket rather than spanning across multiple tags.


# 4. **URL Removal**

### Consider the text:
### For more details visit https://phitron.io or www.phitron.io


### Question:
  - ***Explain the regex:***

    `r'https?://\S+|www\.\S+'`

    ***Ans:*** It matches and removes standard web links starting with http, https, or www.

  - ***What does `https?`, `\S+` and `www\.` mean?***

    ***Ans:***
    - `https?`: Matches "http" with an optional "s".
    - `\S+`: Matches one or more non-whitespace characters following the protocol.
    - `www\.`: Matches the literal string "www." prefix.


# 5. **Punctuation Removal**

### Consider the code:
### `text.translate(str.maketrans('', '', string.punctuation))`


### Question:
  - ***What is the role of string.punctuation?***

    ***Ans:*** `string.punctuation` provides a built-in string containing all standard punctuation characters.

  - ***Why are the first two arguments of maketrans() empty strings?***

    ***Ans:*** The first two arguments are empty because we are deleting characters instead of mapping replacements.

  - ***What would happen if punctuation is not removed before tokenization?***

    ***Ans:*** Attached symbols become part of tokens, creating incorrect unique words like "Dhaka!" instead of "Dhaka".


# 6. **Stopword Removal**

### Consider the sentence:
### I am studying Machine Learning in the university.


### Question:
  - ***What are stopwords?***

    ***Ans:*** Stopwords are common, low-information words like "the", "is", and "at" that carry little analytical meaning.

  - ***Which words would likely be removed?***

    ***Ans:*** "I", "am", "in", "the".

  - ***Why do we convert stopwords into a set instead of keeping them as a list?***

    ***Ans:*** Converting to a set allows for $O(1)$ constant-time lookups, which is much faster than searching through a list.


# 7. **Tokenization**

### Consider the sentence:
### I am going to visit Dhaka!


### Question:
  - ***What is tokenization?***

    ***Ans:*** Tokenization is the process of breaking down a text into smaller functional units like individual words or sentences.

  - ***What is the difference between `text.split()` and `word_tokenize(text)`?***

    ***Ans:*** `text.split()` splits purely by whitespace, while `word_tokenize()` uses language rules to handle punctuation separately.

  - ***Which one handles punctuation better?***

    ***Ans:*** `word_tokenize(text)` handles punctuation much better.


# 8. **Stemming**

### Consider the words:
### running, studies, playing, happiness


### Question:
  - ***What is stemming?***

    ***Ans:*** Stemming is the rule-based process of chopping off word suffixes to reduce them to a rough root form.

  - ***Apply stemming conceptually to each word.***

    ***Ans:*** `run`, `studi`, `play`, `happi`.

  - ***Why might stemming produce words that are not found in a dictionary?***

    ***Ans:*** It may produce non-dictionary words because it simply strips endings without checking a valid vocabulary dictionary.

### Question:
  - ***What is lemmatization?***

    ***Ans:*** Lemmatization reduces words to their proper dictionary root form (lemma) using vocabulary and morphological analysis.

  - ***Why is lemmatization generally more accurate than stemming?***

# 10. **Complete NLP Pipeline**

### Consider the text:
### OMG!!! I was studying NLP at PHITRON 😊.
### Visit https://phitron.io for more details.


### Question:
  - ***Describe the preprocessing steps in order and show the expected output after each stage.***

    ***Ans:***

    1. **Lowercasing:** `omg!!! i was studying nlp at phitron 😊. visit https://phitron.io for more details.`

    2. **URL removal:** `omg!!! i was studying nlp at phitron 😊. visit for more details.`

    3. **Emoji removal:** `omg!!! i was studying nlp at phitron . visit for more details.`

    4. **Punctuation removal:** `omg i was studying nlp at phitron visit for more details`

    5. **Tokenization:** `['omg', 'i', 'was', 'studying', 'nlp', 'at', 'phitron', 'visit', 'for', 'more', 'details']`

    6. **Stopword removal:** `['omg', 'studying', 'nlp', 'phitron', 'visit', 'details']`

    7. **Stemming:** `['omg', 'studi', 'nlp', 'phitron', 'visit', 'detail']`

    8. **Lemmatization:** `['omg', 'study', 'nlp', 'phitron', 'visit', 'detail']`
