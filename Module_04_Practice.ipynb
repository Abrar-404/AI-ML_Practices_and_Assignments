## 1. Why does the perceptron algorithm use class labels −1 and +1 instead of 0 and 1? Explain how this choice affects the perceptron update rule.

- it uses -1 and 1 to add or deduct the weights. If the result is positive then it add more weights to find the best possible outcome, if negative it deducts the weight to find the best outcome.
If it uses 0 and 1, then it can give inaccurate result or give error. Because if the result is < 0, it will flag it as 0 so there would be no changes of weights but if positive it will add more weights.


## 2. Gradient descent usually updates weights by subtracting the gradient. Why does the perceptron update rule add the term ( eta , y , x ) instead?

- Gradient descent updates weights by subtracting the gradient, and the perceptron's gradient for a misclassified point is −η·y·x (it already has a negative sign in it). So subtracting a negative flips it to positive, giving w = w + η·y·x. From there, y·x just pushes weights up when y is +1 and down when y is −1, with no further sign flips.

## 3. Explain in your own words how the perceptron learning rule can be seen as a form of gradient descent. Also mention one limitation of the perceptron loss function.

- The perceptron learning rule can be seen as a form of gradient descent because perceptron first take input and add weights to it so it can find the best possible feature or data. It looks for the missclassified data and update it to find the accurate result. Gradient descent also helps to find the best weights by tuning its derivatives. So it can be said that the perceptron is a form of gradient descent.

- One of the biggest limitation of perceptron is it can only separate linearly separable data. If any non-linear data occur, it doesn't perform well.
