# NLP Pipeline Questions

# Data Acquisition

## 1. What do you understand by Data Acquisition?
### ***Ans:*** Data acquisition is the first step of the NLP pipeline. It means collecting the raw text data needed to solve a problem.

## 2. Why do we need to collect data?
### ***Ans:*** Models learn patterns from data. Without enough relevant data, we cannot train, test or improve a model.

## 3. From where can we collect data for an NLP project?
### ***Ans:*** From public datasets (Kaggle, Hugging Face), websites (web scraping), APIs (Twitter/X, news), company records (emails, chats, reviews), surveys, and by creating our own labeled data.

## 4. What makes a dataset good for solving a problem?
### ***Ans:*** The elements that make a dataset good for solving a problem are:

1. **Relevance:** The data matches the problem. For spam detection, you need emails, not news articles.
2. **Quality and accuracy:** The data has few errors, duplicates, or missing values.
3. **Sufficient size:** There are enough examples for the model to learn patterns.
4. **Correct labels:** In supervised tasks, each example has the right label.
5. **Balance:** Every class has a fair number of examples, so the model isn't biased toward one class.
6. **Diversity:** The data covers different writing styles, topics, and situations.
7. **Real-world similarity:** The data looks like what the model will see after deployment.

# Text Preparation

##1.  What do you understand by Text Preparation?
### ***Ans:*** It is the step where raw text is cleaned and converted into a consistent form that a model can work with.

## 2. Why is text cleaning important?
### ***Ans:*** Raw text has noise such as symbols, links, typos and extra spaces. Noise confuses the model and lowers accuracy, so clean text gives better results.

## 3. What cleaning steps would you perform on raw text data?
### ***Ans:*** Convert to lowercase, remove HTML tags, URLs, Regular Expressions, punctuation, numbers (if not needed), emojis and extra spaces, remove duplicates, and fix obvious spelling errors.

## 4. What is text preprocessing?
### ***Ans:*** Preprocessing is a set of standard operations applied after cleaning to prepare text for modeling. Examples are tokenization, stop-word removal, stemming and lemmatization.

## 4. What preprocessing techniques have you learned so far?
### ***Ans:*** Tokenization, Vocab building, extracting sentences, lowercasing, stop-word removal, stemming, lemmatization, and regex-based cleaning.

## 5. When do we need advanced preprocessing techniques?
### ***Ans:*** When the text is complex or messy, such as slang, mixed languages, many spelling mistakes, or when the task needs POS tagging, named entity recognition (NER) or parsing.

# Regular Expressions (Regex)

## 1. What is a Regular Expression (Regex)?
### ***Ans:*** A regex is a special pattern used to search, match and manipulate text.

## 2. Why do we use Regex in NLP?
### ***Ans:*** It lets us find, extract, replace or remove patterns in large text quickly. It is very useful for cleaning and information extraction.

## 3. What is a regex pattern?
### ***Ans:*** A regex pattern is a string made of normal characters and special symbols that describes the kind of text we want to match. Example: \d+ matches one or more digits.

## 4. What is the purpose of symbols like ., *, +, ?, ^, and $?

### ***Ans:*** "." matches any single character. "*" means zero or more of the previous item. "+" means one or more. "?" means zero or one (optional). "^" matches the start of a string. "$" matches the end of a string.

## 5. How can Regex help us find email addresses, phone numbers, or URLs?
### ***Ans:*** We write a pattern for each format. For example, [\w.]+@\w+\.\w+ finds emails, \d{11} finds 11-digit phone numbers, and https?://\S+ finds URLs.

## 5. How can Regex help in text cleaning?
### ***Ans:*** It can remove or replace unwanted parts like URLs, hashtags, mentions, special characters, digits and extra spaces in one step.

# Feature Engineering



## 1. What do you understand by Feature Engineering?
### ***Ans:*** It is the process of converting text into numerical features (vectors) that a machine learning model can understand and learn from.

## 2. Why can't machine learning models work directly with raw text?
### ***Ans:*** Models only work with numbers. Raw text must first be turned into numeric form.

## 3. When would you choose one text representation technique over another?
### ***Ans:*** Use Bag of Words or TF-IDF for small, simple tasks where speed matters and word counts are enough. Use word embeddings (Word2Vec, GloVe) or transformer models (BERT) when meaning, context and word order matter and you have enough data and computing power.



# Modelling

## 1. What do you understand by Modelling?
### ***Ans:*** Modelling means choosing an algorithm and training it on the prepared features so it learns patterns and can make predictions.

## 2. How do you choose a machine learning algorithm for a problem?
### ***Ans:*** Consider the type of problem (classification, regression, generation), data size, available computing power, need for speed, and need for explainability. Start with a simple baseline, then try better models and compare.

## 3. What algorithm would you use for a text classification task and why?
### ***Ans:*** Naive Bayes or Logistic Regression as a fast, simple baseline that works well with TF-IDF. For higher accuracy, use deep learning models such as LSTM or BERT, which capture context better.

## 4. What is the difference between training and testing a model?
### ***Ans:*** Training is when the model learns from known (labeled) data. Testing is when we check how well it performs on new, unseen data.



# Evaluation

## Intrinsic Evaluation

## 1. What do you understand by model evaluation?
### ***Ans:*** It is the process of measuring how well a trained model performs using suitable metrics.

## 2. Why do we evaluate a model?
### ***Ans:*** To check if it works properly, compare different models, find weaknesses, and decide if it is ready to use.

## 3. What is Accuracy?
### ***Ans:*** The number of correct predictions divided by the total number of predictions.

## 4. What is Precision?
### ***Ans:*** Out of all the items the model predicted as positive, the fraction that are truly positive. Precision = TP / (TP + FP).

## 5. What is Recall?
### ***Ans:*** Out of all the actual positive items, the fraction the model correctly found. Recall = TP / (TP + FN).

## 6. What is F1-Score?
### ***Ans:*** The harmonic mean of precision and recall. F1 = 2 x (Precision x Recall) / (Precision + Recall). It gives one balanced score.

## 7. When is Accuracy not a good metric?
### ***Ans:*** When the data is imbalanced. For example, if 95% of emails are not spam, a model that always says 'not spam' gets 95% accuracy but catches no spam. Use precision, recall or F1 instead.

## Extrinsic Evaluation

## 1. What do you understand by Extrinsic Evaluation?
### ***Ans:*** It evaluates the model by its effect on a real-world task or system, not just by test-set scores.

## 2. How can we determine whether a model is useful in a real-world application?
### ***Ans:*** Run it in the real task with real users and check whether it improves outcomes, such as saving time, reducing errors or increasing user satisfaction. Collect user feedback and run A/B tests.

## 3. Can a model have good accuracy but still fail in practice? Why?
### ***Ans:*** Yes. Real data may differ from test data, the metric may not match what users need, the model may be too slow, or it may fail on important rare cases.

# Deployment

## 1. What do you understand by Model Deployment?
### ***Ans:*** Deployment means placing the trained model into a live environment so that real users or systems can use it.

## 2. Why do we deploy a model?
### ***Ans:*** A model only creates value when it is used to solve real problems. Deployment makes its predictions available to users.

## 3. How can users interact with a deployed model?
### ***Ans:*** Through a web application, mobile app, chatbot, or an API that other software can call.

## 4. What challenges might occur after deployment?
### ***Ans:*** Slow response time, handling many users (scaling), high cost, new or unexpected input, integration bugs, and security or privacy issues.

# Monitoring


## 1. What do you understand by Model Monitoring?
### ***Ans:*** It is the process of continuous;y tracking of a deployed model's performance and behavior.

## 2. Why is monitoring important after deployment?
### ***Ans:*** Real-world data changes over time, so model performance can drop without anyone noticing. Monitoring catches problems early.

## 3. What problems can occur if we do not monitor a model?
### ***Ans:*** Accuracy slowly falls, errors and bias grow unnoticed, system failures go undetected, and users lose trust.

## 4. What would you monitor in a deployed NLP system?
### ***Ans:*** Prediction quality (accuracy, precision, recall), changes in input data, response time, error rates, system usage, and user feedback.



# Model Update

## 1. Why do models need updates?
### ***Ans:*** Language, user behavior, topics and data keep changing, so an old model becomes less accurate over time.

## 2. What is model drift?
### ***Ans:*** Model drift is the drop in a model's performance because real-world data or patterns have changed from the data it was trained on.

## 3. When should a model be retrained?
### ***Ans:*** When performance falls below an agreed level, when drift is detected, or when enough new labeled data has been collected.

## 4. How would you decide whether a new model is better than the current one?
### ***Ans:*** Test both models on the same recent data and compare metrics. Then check the new model on real traffic (A/B test) and replace the old one only if it performs better.
