## Describe Primary Key, Composite Primary Key, Foreign Key with example

### *Ans*:
- Primary Key:
  - A Primary Key is a single column chosen to uniquely identify each record in a table. It must contain completely unique values, it cannot contain NULL values, and a table can only have exactly one primary key.

- Composite Key:
  - A Composite Key is a primary key that consists of two or more columns combined together. It is used when a single column is not enough to guarantee uniqueness for a record. Individual columns inside a composite key can have duplicate values across rows, but the combination of all the columns together must be completely unique.

- Foreign Key:
  - A Foreign Key is a column (or group of columns) in one table that points to the Primary Key of another table. It establishes a relational link and enforces referential integrity, meaning you cannot add a record to the child table unless the referenced record exists in the parent table. nlike primary keys, a foreign key can accept duplicate values (one-to-many relationships) and can also accept NULL values depending on your database design.

## Write five numeric type Data Types of MySQL

### *Ans*:
1. **CHAR(size)**:
   - A fixed-length string (can contain letters, numbers, and special characters). size specifies the column length in characters, from 0 to 255. Default size is 1

2. **VARCHAR(size)**:
   - A variable-length string (can contain letters, numbers, and special characters). size specifies the maximum column length in characters, from 0 to 65535.

3. **BINARY(size)**:
   - Similar to CHAR(), but stores binary byte strings. size specifies the column length in bytes. Default size is 1

4. **SET(val1, val2, val3, ...)**:
   - A string object that can have zero or more values, chosen from the list of possible values (val1, val2, val3,..). You can list up to 64 values

5. **VARBINARY(size)**:
   - Similar to VARCHAR(), but stores binary byte strings. size specifies the maximum column length in bytes.
